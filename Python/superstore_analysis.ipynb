```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../Data/Raw/Sample - Superstore.csv", encoding='latin1')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
df.info()
```

```python
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
df['Month Name'] = df['Order Date'].dt.month_name()
df['Profit Margin'] = (df['Profit'] / df['Sales']) * 100
```

VISUAL EXPLORATORY ANALYSIS

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Revenue by Category Chart

```python
category_sales = df.groupby('Category')['Sales'].sum()
category_sales.plot(kind = 'bar')

plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.savefig('sales_by_category.png', bbox_inches = 'tight')

plt.show()
```

Regional Sales Chart

```python
region_sales = df.groupby('Region')['Sales'].sum()

region_sales.plot(kind = 'bar')

plt.title("Sales by Region")
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.savefig('sales_by_region.png', bbox_inches='tight')

plt.show()
```

Monthly Revenue Trend

```python
monthly_sales = df.groupby('Month Name')['Sales'].sum()
monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.savefig("monthly_sales_trend.png", bbox_inches="tight")

plt.show()
```

Top 10 Products

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

top_products.plot(kind = 'barh')

plt.title('Top 10 Products by Sales')
plt.xlabel('Sales')

plt.show()
```

FIXING MONTH ORDER

```python
month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

monthly_sales = df.groupby('Month Name')['Sales'].sum().reindex(month_order)
monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.show()
```

ADVANCED BUSINESS ANALYSIS

Profit Analysis by Category

```python
category_profit = df.groupby('Category')['Profit'].sum()

category_profit.plot(kind = 'bar')

plt.title('Profit by Category')
plt.xlabel('Category')
plt.ylabel('Profit')

plt.show()
```

Regional Profit Analysis

```python
region_profit = df.groupby('Region')['Profit'].sum().sort_values()

region_profit.plot(kind = 'bar')

plt.title('Profit by Region')
plt.xlabel('Region')
plt.ylabel('Profit')

plt.show()
```

Discount vs Profit

```python
plt.scatter(df['Discount'], df['Profit'])

plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.savefig("discount_profit_analysis.png", bbox_inches="tight")

plt.show()
```

Customer Segment Analysis

```python
segment_sales = df.groupby('Segment')['Sales'].sum()

segment_sales.plot(kind='pie', autopct='%1.1f%%')

plt.title('Sales by Customer Segment')
plt.ylabel('')

plt.show()
```

Shipping Mode Analysis

```python
ship_mode_sales = df.groupby('Ship Mode')['Sales'].sum()

ship_mode_sales.plot(kind = 'bar')

plt.title('Sales by Shipping Mode')
plt.xlabel('Shipping Mode')
plt.ylabel('Sales')

plt.show()
```
